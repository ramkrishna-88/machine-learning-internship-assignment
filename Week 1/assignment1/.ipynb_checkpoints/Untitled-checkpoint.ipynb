### Titanic Data Cleaning

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
df = pd.read_csv("D:\Titanic-Dataset.csv")
print("Dataset loaded successfully!")
df.head()
```

```python
print("Shape of dataset:", df.shape)
```

```python
print("Columns:")
print(df.columns)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.describe(include="all")
```

```python
print(df.isnull().sum())
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

###### Handling missing value


```python
print("Missing Age values:", df["Age"].isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
print("Missing Age values after imputation:", df["Age"].isnull().sum())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
print(df["Embarked"].isnull().sum())
```

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")
```

```python
print(df.isnull().sum())
```

##### Encode Sex using LabelEncoder

```python
label_encoder = LabelEncoder()
```

```python
df["Sex"] = label_encoder.fit_transform(df["Sex"])
df[["Sex"]].head()
```

```python
print(label_encoder.classes_)
```

###### Encode using OneHotEncoder

```python
print(df["Embarked"].unique())
```

```python
one_hot_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)
embarked_encoded = one_hot_encoder.fit_transform(
    df[["Embarked"]]
)
```

```python
encoded_columns = one_hot_encoder.get_feature_names_out(["Embarked"])

print(encoded_columns)
```

```python
embarked_df = pd.DataFrame(
    embarked_encoded,
    columns=encoded_columns,
    index=df.index
)

embarked_df.head()
```

```python
df = pd.concat([df, embarked_df], axis=1)

# Drop Embarked only if it exists
if "Embarked" in df.columns:
    df.drop("Embarked", axis=1, inplace=True)

print(df.head())
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["Age"], bins=30, kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["Age"])

plt.title("Age Distribution - Boxplot")
plt.xlabel("Age")

plt.show()
```

```python
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())
```
